```python
from ultralytics import YOLO

# 1. Cargar el modelo preentrenado (descargará los pesos automáticamente)
# Puedes cambiar 'yolov8m.pt' por 'yolo11m.pt' si quieres probar la versión más reciente
model = YOLO('yolov8m.pt') 

# 2. Configurar y lanzar el entrenamiento
results = model.train(
    data=f"{dataset.location}/data.yaml", # Ruta automática al dataset descargado
    epochs=100,               # 100 es un buen inicio para ver cómo converge
    imgsz=640,                # Resolución. Si los paneles se ven muy pequeños, súbelo a 1024
    batch=16,                 # Si cambias imgsz a 1024, baja el batch a 8 para que la T4 no se quede sin memoria (CUDA Out of Memory)
    device=0,                 # Indica que use la GPU T4
    patience=20,              # Si el modelo no mejora en 20 epochs, se detiene solo para ahorrar tiempo
    
    # --- Hiperparámetros clave para Drones ---
    degrees=15.0,             # Rotación aleatoria (útil porque el dron no siempre está alineado)
    scale=0.5,                # Aleja o acerca la imagen para simular cambios de altitud del dron
    perspective=0.0001,       # Simula inclinación de la cámara
    fliplr=0.5,               # Volteo horizontal (50% de probabilidad)
    flipud=0.5,               # Volteo vertical (50% de probabilidad, útil en vistas cenitales)
    
    # --- Guardado de resultados ---
    project='Deteccion_Techos_PV',
    name='Prueba_YOLO_Medium'
)
```
